```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1. Read an image from disk using OpenCV
# cv2.imread loads the image as a NumPy ndarray
bgr_img = cv2.imread("annotated_photo.jpg")

print("Type:", type(bgr_img))
print("Shape (Height, Width, Channels):", bgr_img.shape)
print("Data type:", bgr_img.dtype)  # uint8

# 2. Fix Color Order: BGR -> RGB
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)

# 3. View in Jupyter using Matplotlib
plt.figure(figsize=(5, 3))
plt.imshow(rgb_img)
plt.title("Proper RGB Display in Notebook")
plt.axis("off")
plt.show()
```

```python
# Convert to single-channel Grayscale
gray_cv = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2GRAY)
print("Grayscale shape (no color channels):", gray_cv.shape)

# Convert to HSV (useful for color-based object tracking)
hsv_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2HSV)
```

```python
# Any pixel brighter than 180 becomes 255 (white); otherwise 0 (black)
threshold_val = 180
max_val = 255
ret, binary_mask = cv2.threshold(gray_cv, threshold_val, max_val, cv2.THRESH_BINARY)

plt.figure(figsize=(5, 3))
plt.imshow(binary_mask, cmap="gray")
plt.title("Binary Threshold Mask")
plt.axis("off")
plt.show()
```

```python
# Canny takes two thresholds for its hysteresis procedure:
# threshold1 = minVal (weak edges)
# threshold2 = maxVal (strong edges)
edges = cv2.Canny(image=gray_cv, threshold1=50, threshold2=150)

plt.figure(figsize=(5, 3))
plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

```python
# 1. Find continuous contours from the edge/binary map
# cv2.RETR_EXTERNAL: retrieves only the extreme outer contours
# cv2.CHAIN_APPROX_SIMPLE: compresses horizontal, vertical, and diagonal segments
contours, hierarchy = cv2.findContours(
    edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
)

print(f"Detected {len(contours)} contours")

# 2. Make a copy of our RGB image to draw on
annotated = rgb_img.copy()

# 3. Iterate over contours and draw green bounding boxes around them
for c in contours:
    # Get bounding coordinates: x, y, width, height
    x, y, w, h = cv2.boundingRect(c)

    # Filter out tiny noise specks
    if w > 20 and h > 20:
        # cv2.rectangle(image, pt1, pt2, color, thickness)
        cv2.rectangle(annotated, (x, y), (x + w, y + h), (0, 255, 0), 2)

plt.figure(figsize=(5, 3))
plt.imshow(annotated)
plt.title("Contours with Bounding Boxes")
plt.axis("off")
plt.show()
```
