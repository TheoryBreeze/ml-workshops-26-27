```python
from PIL import Image, ImageDraw, ImageFilter, ImageFont

# Step 0: Create a blank RGB image (Width: 400, Height: 300) with a sky-blue background
img = Image.new(mode="RGB", size=(400, 300), color=(135, 206, 235))

# Save it to disk so we can demonstrate loading files
img.save("sample_photo.jpg")
```

```python
# 1. Open an image file
img = Image.open("sample_photo.jpg")

# 2. Inspect properties
print(f"Dimensions (width, height): {img.size}")
print(f"Color Mode: {img.mode}")  # RGB, RGBA, L
print(f"File Format: {img.format}")  # JPEG, PNG, etc.
```

```python
# A. Resize (Pass explicit new size tuple: width, height)
resized_img = img.resize((200, 150))

# B. Crop a region: box = (left, top, right, bottom)
# Crops a 100x100 square from the middle
crop_box = (100, 50, 300, 250)
cropped_img = img.crop(crop_box)

# C. Rotate & Flip
rotated_img = img.rotate(45, expand=True)  # expand=True fits entire rotated canvas
flipped_img = img.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
```

```python
# Create a drawing context bound to the image
draw = ImageDraw.Draw(img)

# Draw a red rectangle (e.g., simulating a bounding box)
# box coordinates: [x0, y0, x1, y1]
draw.rectangle([50, 50, 200, 180], outline="red", width=4)

# Draw a yellow sun circle: [x0, y0, x1, y1]
draw.ellipse([280, 30, 360, 110], fill="gold", outline="orange", width=2)

# Draw text
draw.text((60, 60), "Hello Python!", fill="black")

# Save result
img.save("annotated_photo.jpg")
```

```python
# Convert to Grayscale (Mode "L")
gray_img = img.convert("L")

# Apply built-in convolution filters
blurred_img = img.filter(ImageFilter.GaussianBlur(radius=3))
edge_img = gray_img.filter(ImageFilter.FIND_EDGES)
```
